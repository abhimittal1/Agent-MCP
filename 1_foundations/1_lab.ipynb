```python
from openai import OpenAI
openai = OpenAI()
```

```python
messages = [{"role": "user", "content": "What is 2+2?"}]
```

```python
response = openai.chat.completions.create(
    model="gpt-4.1-nano",
    messages=messages
)

print(response.choices[0].message.content)
```

```python
question = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question}]
```

```python
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)

question = response.choices[0].message.content

print(question)
```

```python
messages = [{"role": "user", "content": question}]
```

```python
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=messages
)

answer = response.choices[0].message.content
print(answer)
```

```python
from IPython.display import Markdown, display

display(Markdown(answer))
```

```python
question = "I want you to explore the business area that might be the most profitable in the next 10 years. Please provide a detailed analysis of the top 3 business areas, including potential growth, challenges, and key players in each area."
message = [{"role": "user", "content": question}]
```

```python
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=message
)   
answer = response.choices[0].message.content
display(Markdown(answer))
```

```python
message = [{"role": "user", "content": answer}]
response = openai.chat.completions.create(
    model="gpt-4.1-mini",
    messages=message
)
answer = response.choices[0].message.content
display(Markdown(answer))
```
